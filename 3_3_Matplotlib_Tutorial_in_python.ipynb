#**Matplotlib:**
    *   Useful for making plots








```python
# importing Matplotlib library
import matplotlib.pyplot as plt

# import numpy to get data for our plots
import numpy as np
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)
z = np.cos(x)
```

```python
print(x)
```

```python
print(y)
```

```python
print(z)
```

## Plotting the data

```python
# sine wave
plt.plot(x, y)
plt.show()
```

```python
# Cosin wave
plt.plot(x, z)
plt.show()
```

```python
# Adding title, X-axis and Y-axis labels
plt.plot(x, y)
plt.xlabel('angle')
plt.ylabel('sine wave')
plt.title('sine wave')
plt.show()
```

# Parabola

```python
x = np.linspace(-10, 10, 20)    # give 20 equal space values from range of -10 to 10
y = x**2
plt.plot(x, y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('parabloa')
plt.show()
```

```python
plt.plot(x, y, 'r+')        # r+ : Red color with + symbols in it

#plt.plot (x, y, 'g.')       g. : green color with dotted curve line in it

plt.show()
```

```python
plt.plot(x, y, 'rx')
plt.show()
```

```python
# multiple plot in single graph
x = np.linspace(-5, 5, 50)
plt.plot(x, np.sin(x), 'g-')
plt.plot(x, np.cos(x), 'r--')
plt.show()
```

#Bar plot

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
languages = ['English', 'french', 'Spanich', 'latin', 'German']
people = [100, 50, 150, 40, 70]
ax.bar(languages, people)
plt.xlabel('LANGUAGES')
plt.ylabel('NUMBER OF PEOPLE')
```

#Pie chart

```python
fig1 = plt.figure()
ax = fig1.add_axes([0,0,1,1])
languages = ['English', 'Nepali', 'German', 'Latin', 'France' ]
people = [100, 200, 50, 150, 120]
ax.pie(people, labels=languages, autopct='%1.1f%%')
plt.show()
```

#Scatter plot

```python
x = np.linspace(0, 10, 30)
y = np.sin(x)
z = np.cos(x)

fig2 = plt.figure()
ax = fig2.add_axes([0,0,1,1])
ax.scatter(x,y, color= 'r')
ax.scatter(x,z, color= 'b')
plt.show()
```

#3D Scatter plot

```python
fig3 = plt.figure()
ax = plt.axes(projection = '3d')
z = 20* np.random.random(100)
x = np.sin(z)
y = np.cos(z)
ax.scatter(x, y, z, c=z, cmap = 'Blues')
plt.show()
```
