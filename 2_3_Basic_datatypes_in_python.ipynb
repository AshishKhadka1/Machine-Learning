# Slicing

```python
my_string = "Programming"
print(my_string[1:5])       #values form 1 to 5-1 will be sliced

print(my_string[0:10:2])    # 0 → start at index 0 (first character)
                            # 10 → go up to index 10 (but not including 10)
                            # 2 → take every 2nd character
```

# String Concatenation
