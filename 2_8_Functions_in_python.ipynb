#**Functions**
function is a block of code that can be reused in a program

**Factorial of a Number**
factorial of a given number is the product of all the positive integers that is less than or equal to the given number.

Factorial of 5 : 5 x 4 x 3 x 2 x 1 = 120

```python
number = int(input(f"Enter a number to find it's factorial : "))
factorial = 1
if number == 0:
  print(' The factorial of 0 is 1 ')
else:
  for i in range(1, number + 1):
    factorial = factorial * i
  print('Factorial of ', number, ' is ', factorial)
```

**Factorial Function**


```python
def factorial_value(num):
  factorial = 1

  if num == 0:
    return factorial
  else:
    for i in range(1, num + 1):
      factorial = factorial * i
    return factorial
```

```python
print(factorial_value(4))
print(factorial_value(0))
```
