#1. Arithmetic Operators

```python
num_1 = int(input("Enter a First number: "))
num_2 = int(input("Enter a Second Number: "))

#Addition
sum = num_1 +num_2
print(f'Sum of {num_1} and {num_2} is = ', sum)

#Subtraction
diff = num_1 - num_2
print(f"Difference of {num_1} with {num_2} is  = ", diff)

#Division
quo = num_1 / num_2
print(f"Quotient of {num_1} with {num_2} is = ", quo)

#Multiplication
pro = num_1 * num_2
print(f"Multiplication of {num_1} and {num_2} is = ", pro)

#Exponent / Power
exp = num_1 **num_2
print(f"Power of {num_1} and {num_2} is = ", exp)

#Modulus
mod = num_1 % num_2
print(f"Remainder of {num_1} with {num_2} is = ", mod)
```

#3. Comparison Operator

```python
a = 5
b = 10

print (a == b)
print (a > b)
print (a < b)
print (a >= b)
print (a <= b)
print ( a!= b)    #not equal to
```

#4. Logical Operator

and

or

not

```python
a = 10
print (a>20 and a>5)        #Returns True only if both conditions are True.
print ( a<10 or a>2)        #Returns True if at least one condition is True.
print ( not(a<10 and a>2))  #Inverts the Boolean value: True → False, False → True.
```

#6. Membership Operator

```python
a = 10
b = 11

c = [2, 4, 5, 10]
print (b in c)
print (a in c)

print (a not in c)
print (b not in c)
```
