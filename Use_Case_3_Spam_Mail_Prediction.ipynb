**Importing the Dependencies**



```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# TfidfVectorizer: convert Text data into Numerical values
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

**Data Collection & Pre-Processing**

```python
# loading the data from csv file to a pandas Dataframe
raw_mail_data = pd.read_csv('/content/mail_data.csv')
```

```python
print(raw_mail_data)
```

```python
# replace the null values with a null string
mail_data = raw_mail_data.where((pd.notnull(raw_mail_data)),'')
```

```python
# printing the first 5 rows of the dataframe
mail_data.head()
```

```python
# checking the number of rows and columns in the dataframe
mail_data.shape
```

**Label Encoding**

```python
# label spam mail as 0;  ham mail as 1;

mail_data.loc[mail_data['Category'] == 'spam', 'Category',] = 0
mail_data.loc[mail_data['Category'] == 'ham', 'Category',] = 1
```

* spam - 0

* ham - 1

```python
# separating the data as texts and label

X = mail_data['Message']

Y = mail_data['Category']
```

```python
print(X)
```

```python
print(Y)
```

**Splitting the data into training data & test data**

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=3)
```

```python
print(X.shape)
print(X_train.shape)
print(X_test.shape)
```

**Feature Extraction**

```python
# convert textual data to Feature Vectors that can be used as input to Logistic Regression Model
# min_df=2 means ignore the words that appears in less than 2 documents
feature_extraction = TfidfVectorizer(min_df = 2 , stop_words = 'english', lowercase = True)

X_train_features = feature_extraction.fit_transform(X_train)
X_test_features = feature_extraction.transform(X_test)

# convert Y_train and Y_test values to integers
Y_train = Y_train.astype('int')
Y_test = Y_test.astype('int')
```
