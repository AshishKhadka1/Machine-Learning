# **Types of Object in Python:**
1. Immutable Objects
2. Mutable Objects

**Immutable Objects**
1. int
2. float
3. string
4. bool
5. tuple

**Mutable Objects**
1. List
2. Set
3. Dictionary

#**List**

```python
my_list = [1, 2, 3.1, 'English', True]
print(my_list)
type(my_list)
# List are Mutable -> Changeable

print(my_list[2])
print(my_list[4])
```

```python
# list allow dublicate values
list_1 = [1, 2, 3, 1, 2 ,3 , 6]
print(list_1)
```

```python
# Initiating an Empty list
list_2 = []
print(list_2)
```

```python
# Append
list_2.append(5)
print(list_2)
```

```python
# Delete an item in the list
list_2 = [2, 3, 1.8, 'Learn', True, 0]
print(list_2)
del list_2[3]
print(list_2)
```

```python
# Join two lists
list_3 = [1,2,3,4,5]
list_4 = [1,2,3,4,5]

list_5 = list_3 + list_4
print(list_5)
```

#**Tuple**

```python
# Tuple allows multiple data types
tuple_1 = (1, 3, 'Hello world', 5.5 , False)
print(tuple_1)
type(tuple_1)
```

```python
# Converting a list into a tuple
my_list = [3, 4, 6, 7 ,9]
print(my_list)

tuple = tuple(my_list)
print(tuple)
```

```python
print(tuple[2])
```

Tuples are Immutable -> Unchangeable

```python
tuple.append(9)
```

```python
print(len(tuple))
```

#Set

```python
# set --> Curley Brackets
my_set = {1, 3, 4, 5, 6, 8}
print(my_set)
type(my_set)
```

```python
# Object does not support indexing
print(my_set[0])
```

```python
# Convert a list into a Set
list_5 = [2,3,5,6,78]
X = set(list_5)
print(X)
```

```python
# Set doesnot allow multiple values
set_3 = {1, 2, 1, 2 ,3, 4, 5,3, 5}
print(set_3)
```

#**Dictionary**

Key - Value Pair


```python
my_dictionary = {'name': 'Ashish', 'cast': 'Khadka', 'Age': 30, 'Country': 'Nepal'}
print(my_dictionary)
type(my_dictionary)
```

```python
print(my_dictionary['cast'])
```

```python
# Dictionary doesnot allow dublicate values
my_dictionary2 = {'name': 'Ashish', 'cast': 'Khadka','cast': 'khadka','Age': 23, 'Age': 30, 'Country': 'Nepal'}
print(my_dictionary2)
```
