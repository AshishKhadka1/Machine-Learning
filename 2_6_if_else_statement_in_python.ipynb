#**`Simple if else Statement`**




```python
a = 10
b = 20

if (a>b):
  print("a is greatest number")
else:
    print("b is greatest number")
```

#**if elif else Statement**

```python
a = 10
b = 20
c = 15

if a > b and a > c:
    print('a is the greatest number')
elif b > a and b > c:
    print('b is the greatest number')
else:
    print('c is the greatest number')
```
